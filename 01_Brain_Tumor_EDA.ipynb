```python
import hashlib
import os

valid_extensions = ('.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff')

import hashlib
import os

def get_file_hash(file_path):
    hash_md5 = hashlib.md5()

    with open(file_path, "rb") as f:
        for chunk in iter(lambda: f.read(4096), b""):
            hash_md5.update(chunk)

    return hash_md5.hexdigest()


image_hashes = {}
duplicate_files = []

for split in ['Training', 'Testing']:

    for class_name in classes:

        folder = os.path.join(dataset_path, split, class_name)

        for filename in os.listdir(folder):

            if filename.lower().endswith(valid_extensions):

                image_path = os.path.join(folder, filename)

                try:
                    file_hash = get_file_hash(image_path)

                    if file_hash in image_hashes:
                        duplicate_files.append(
                            (image_path, image_hashes[file_hash])
                        )
                    else:
                        image_hashes[file_hash] = image_path

                except Exception:
                    pass


print("DUPLICATE IMAGE CHECK")
print("=" * 50)

print("Total duplicate files found:",
      len(duplicate_files))

if len(duplicate_files) == 0:
    print("✅ No exact duplicate files found.")
else:
    print("\nSome duplicate pairs:")
    for duplicate in duplicate_files[:10]:
        print("\nDuplicate:", duplicate[0])
        print("Original :", duplicate[1])
```

```python
import os
import hashlib
from collections import defaultdict

train_path = os.path.join(dataset_path, "Training")
test_path = os.path.join(dataset_path, "Testing")

def get_image_hash(file_path):
    with open(file_path, "rb") as f:
        return hashlib.md5(f.read()).hexdigest()

hash_locations = defaultdict(list)

# Scan Training
for class_name in classes:
    folder = os.path.join(train_path, class_name)

    for file in os.listdir(folder):
        if file.lower().endswith(('.jpg', '.jpeg', '.png', '.bmp')):
            path = os.path.join(folder, file)
            file_hash = get_image_hash(path)

            hash_locations[file_hash].append(
                ("Training", class_name, file)
            )

# Scan Testing
for class_name in classes:
    folder = os.path.join(test_path, class_name)

    for file in os.listdir(folder):
        if file.lower().endswith(('.jpg', '.jpeg', '.png', '.bmp')):
            path = os.path.join(folder, file)
            file_hash = get_image_hash(path)

            hash_locations[file_hash].append(
                ("Testing", class_name, file)
            )

# Find duplicates
duplicate_groups = {
    h: locations
    for h, locations in hash_locations.items()
    if len(locations) > 1
}

print("Total duplicate groups:", len(duplicate_groups))
```

```python
cross_split_duplicates = []

for file_hash, locations in duplicate_groups.items():

    has_training = any(x[0] == "Training" for x in locations)
    has_testing = any(x[0] == "Testing" for x in locations)

    if has_training and has_testing:
        cross_split_duplicates.append(locations)

print("Duplicate groups appearing in BOTH Training and Testing:",
      len(cross_split_duplicates))
```

```python
import tensorflow as tf

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

def load_image(path, label):
    image = tf.io.read_file(path)
    image = tf.image.decode_image(
        image,
        channels=3,
        expand_animations=False
    )
    image = tf.image.resize(image, IMG_SIZE)
    image = tf.cast(image, tf.float32)

    return image, label


# Convert labels to numbers
train_labels = train_df['label'].map(class_to_index).values
val_labels = val_df['label'].map(class_to_index).values

train_paths = train_df['filepath'].values
val_paths = val_df['filepath'].values


train_ds = tf.data.Dataset.from_tensor_slices(
    (train_paths, train_labels)
)

val_ds = tf.data.Dataset.from_tensor_slices(
    (val_paths, val_labels)
)


train_ds = train_ds.map(
    load_image,
    num_parallel_calls=tf.data.AUTOTUNE
)

val_ds = val_ds.map(
    load_image,
    num_parallel_calls=tf.data.AUTOTUNE
)


train_ds = train_ds.shuffle(1000, seed=42)
train_ds = train_ds.batch(BATCH_SIZE)
train_ds = train_ds.prefetch(tf.data.AUTOTUNE)

val_ds = val_ds.batch(BATCH_SIZE)
val_ds = val_ds.prefetch(tf.data.AUTOTUNE)


print("Training batches:", tf.data.experimental.cardinality(train_ds).numpy())
print("Validation batches:", tf.data.experimental.cardinality(val_ds).numpy())
```

```python
def load_image(filepath, label):
    image = tf.io.read_file(filepath)
    image = tf.image.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image = tf.image.resize(image, IMG_SIZE)
    image = tf.cast(image, tf.float32)

    return image, label
```

```python
train_paths = train_df['filepath'].values
train_labels = train_df['label'].values

val_paths = val_df['filepath'].values
val_labels = val_df['label'].values

train_ds = tf.data.Dataset.from_tensor_slices(
    (train_paths, train_labels)
)

val_ds = tf.data.Dataset.from_tensor_slices(
    (val_paths, val_labels)
)

train_ds = train_ds.shuffle(
    buffer_size=len(train_df),
    seed=42
)

train_ds = train_ds.map(
    load_image,
    num_parallel_calls=tf.data.AUTOTUNE
)

val_ds = val_ds.map(
    load_image,
    num_parallel_calls=tf.data.AUTOTUNE
)

train_ds = train_ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
val_ds = val_ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

print("TensorFlow datasets created successfully.")
```

```python
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomRotation(0.05),
    tf.keras.layers.RandomZoom(0.10),
    tf.keras.layers.RandomTranslation(
        height_factor=0.05,
        width_factor=0.05
    ),
    tf.keras.layers.RandomContrast(0.10)
], name="data_augmentation")

print("Data augmentation created.")
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import ResNet50

NUM_CLASSES = 4

# Load pretrained ResNet50
base_model = ResNet50(
    weights='imagenet',
    include_top=False,
    input_shape=(224, 224, 3)
)

# Freeze the pretrained layers
base_model.trainable = False

# Build our classifier
inputs = layers.Input(shape=(224, 224, 3))

x = data_augmentation(inputs)

# ResNet50 preprocessing
x = tf.keras.applications.resnet50.preprocess_input(x)

x = base_model(x, training=False)

x = layers.GlobalAveragePooling2D()(x)

x = layers.Dropout(0.3)(x)

outputs = layers.Dense(
    NUM_CLASSES,
    activation='softmax'
)(x)

resnet_model = models.Model(
    inputs,
    outputs,
    name="NeuroVision_ResNet50"
)

resnet_model.summary()
```

```python
resnet_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-3
    ),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("ResNet50 model compiled successfully.")
```

```python
def make_gradcam(filepath):

    # Load image
    image = Image.open(filepath).convert("RGB")
    image = image.resize((224, 224))

    img_array = np.array(image).astype("float32")
    img_array = np.expand_dims(img_array, axis=0)

    # ResNet50 preprocessing
    processed_image = tf.keras.applications.resnet50.preprocess_input(
        img_array
    )

    with tf.GradientTape() as tape:

        # Get convolutional output and ResNet features
        conv_outputs, features = grad_model(
            processed_image,
            training=False
        )

        # Pass features through our classification head
        x = resnet_model.get_layer("global_average_pooling2d")(features)
        x = resnet_model.get_layer("dropout")(x, training=False)
        predictions = resnet_model.get_layer("dense")(x)

        predicted_class = tf.argmax(predictions[0])
        class_score = predictions[:, predicted_class]

    # Calculate gradients
    grads = tape.gradient(class_score, conv_outputs)

    # Average gradients
    pooled_grads = tf.reduce_mean(
        grads,
        axis=(0, 1, 2)
    )

    # Remove batch dimension
    conv_outputs = conv_outputs[0]

    # Weighted combination
    heatmap = tf.reduce_sum(
        conv_outputs * pooled_grads,
        axis=-1
    )

    # ReLU
    heatmap = tf.maximum(heatmap, 0)

    # Normalize
    max_value = tf.reduce_max(heatmap)

    if max_value > 0:
        heatmap = heatmap / max_value

    return image, predictions[0].numpy(), heatmap.numpy()
```

```python
def make_gradcam(filepath):

    # Load original image
    image = Image.open(filepath).convert("RGB")
    image = image.resize((224, 224))

    img_array = np.array(image).astype("float32")
    img_array = np.expand_dims(img_array, axis=0)

    # ResNet50 preprocessing
    processed_image = tf.keras.applications.resnet50.preprocess_input(
        img_array
    )

    with tf.GradientTape() as tape:

        # Get convolutional features
        conv_outputs, base_outputs = grad_model(
            processed_image,
            training=False
        )

        # Recreate the classification head
        x = resnet_model.layers[-3](base_outputs)
        x = resnet_model.layers[-2](x)
        predictions = resnet_model.layers[-1](x)

        predicted_class = tf.argmax(predictions[0])
        class_score = predictions[:, predicted_class]

    # Gradients
    grads = tape.gradient(
        class_score,
        conv_outputs
    )

    # Average gradients
    pooled_grads = tf.reduce_mean(
        grads,
        axis=(0, 1, 2)
    )

    # Remove batch dimension
    conv_outputs = conv_outputs[0]

    # Weighted feature maps
    heatmap = tf.reduce_sum(
        conv_outputs * pooled_grads,
        axis=-1
    )

    # ReLU
    heatmap = tf.maximum(heatmap, 0)

    # Normalize
    max_value = tf.reduce_max(heatmap)

    if max_value > 0:
        heatmap = heatmap / max_value

    return image, predictions[0].numpy(), heatmap.numpy()
```

```python
def make_gradcam(filepath):

    # Load image
    image = Image.open(filepath).convert("RGB")
    image = image.resize((224, 224))

    img_array = np.array(image).astype("float32")
    img_array = np.expand_dims(img_array, axis=0)

    # ResNet50 preprocessing
    processed_image = tf.keras.applications.resnet50.preprocess_input(
        img_array
    )

    with tf.GradientTape() as tape:

        # Get convolutional output and ResNet features
        conv_outputs, features = grad_model(
            processed_image,
            training=False
        )

        # Pass features through our classification head
        x = resnet_model.get_layer("global_average_pooling2d")(features)
        x = resnet_model.get_layer("dropout")(x, training=False)
        predictions = resnet_model.get_layer("dense")(x)

        predicted_class = tf.argmax(predictions[0])
        class_score = predictions[:, predicted_class]

    # Calculate gradients
    grads = tape.gradient(class_score, conv_outputs)

    # Average gradients
    pooled_grads = tf.reduce_mean(
        grads,
        axis=(0, 1, 2)
    )

    # Remove batch dimension
    conv_outputs = conv_outputs[0]

    # Weighted combination
    heatmap = tf.reduce_sum(
        conv_outputs * pooled_grads,
        axis=-1
    )

    # ReLU
    heatmap = tf.maximum(heatmap, 0)

    # Normalize
    max_value = tf.reduce_max(heatmap)

    if max_value > 0:
        heatmap = heatmap / max_value

    return image, predictions[0].numpy(), heatmap.numpy()
```
